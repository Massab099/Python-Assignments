### Q14. Weak Password Rejection

A website requires passwords to contain at least 8 characters, including at least one digit. A password that does not meet the rules must be rejected.

Create a custom exception named `WeakPasswordError`. Ask the user for a password, check both rules, raise the custom exception when a rule fails, and handle it to explain what needs to be corrected. Do not display the actual password in the success message.

```python
class WeakPasswordError(Exception):
    pass


try:
    password = input("Enter your password: ")

    if len(password) < 8:
        raise WeakPasswordError("Password must contain at least 8 characters.")

    if not any(char.isdigit() for char in password):
        raise WeakPasswordError("Password must contain at least one digit.")

    print("Password accepted successfully!")

except WeakPasswordError as e:
    print("Weak Password:", e)
```

```python
class InvalidTransferError(Exception):
    pass


class InsufficientFundsError(Exception):
    pass


sender_balance = 50000
receiver_balance = 20000

try:
    amount = float(input("Enter transfer amount: "))

    if amount <= 0:
        raise InvalidTransferError("Transfer amount must be positive.")

    if sender_balance == receiver_balance:
        pass

    if amount > sender_balance:
        raise InsufficientFundsError("Sender has insufficient funds.")

    sender_balance = sender_balance - amount
    receiver_balance = receiver_balance + amount

    print("Transfer successful!")

except InvalidTransferError as e:
    print("Transfer Error:", e)

except InsufficientFundsError as e:
    print("Insufficient Funds:", e)

except ValueError:
    print("Invalid input! Please enter a numeric amount.")

finally:
    print("\nFinal Account Balances:")
    print("Sender Balance:", sender_balance)
    print("Receiver Balance:", receiver_balance)
```

### Q28. Bulk Employee Payroll Processing

A company stores several employee records in a list. Each record contains an employee name, hours worked, and hourly rate. One invalid record should not stop the payroll calculation for all other employees.

Create a custom exception named `InvalidPayrollRecordError`. Validate each employee record, raise this exception when required data is missing or values are invalid, and handle the error inside the record-processing loop. Calculate payroll for valid employees and report which records were skipped and why.

```python
class InvalidPayrollRecordError(Exception):
    pass


employees = [
    {"name": "Ali", "hours": 40, "rate": 500},
    {"name": "Sara", "hours": 35, "rate": 600},
    {"name": "", "hours": 30, "rate": 400},
    {"name": "Ahmed", "hours": -5, "rate": 500},
    {"name": "Ayesha", "hours": 45, "rate": 700},
    {"name": "Usman", "hours": 40, "rate": -200}
]

total_payroll = 0
skipped_records = 0

for employee in employees:

    try:
        name = employee.get("name")
        hours = employee.get("hours")
        rate = employee.get("rate")

        if not name:
            raise InvalidPayrollRecordError("Employee name is missing.")

        if hours is None or rate is None:
            raise InvalidPayrollRecordError("Hours or hourly rate is missing.")

        if hours < 0 or hours > 80:
            raise InvalidPayrollRecordError("Invalid working hours.")

        if rate <= 0:
            raise InvalidPayrollRecordError("Hourly rate must be positive.")

        salary = hours * rate
        total_payroll += salary

        print(name, "- Salary:", salary)

    except InvalidPayrollRecordError as e:
        skipped_records += 1
        print("Record Skipped:", e)

print("\n----- PAYROLL SUMMARY -----")
print("Total Payroll:", total_payroll)
print("Skipped Records:", skipped_records)
```

```python
 class ProductNotFoundError(Exception):
    pass


class InvalidQuantityError(Exception):
    pass


class InsufficientStockError(Exception):
    pass


inventory = {
    "P101": {"name": "Keyboard", "stock": 20},
    "P102": {"name": "Mouse", "stock": 15},
    "P103": {"name": "Monitor", "stock": 10}
}

try:
    product_id = input("Enter product ID: ").upper().strip()
    quantity = int(input("Enter purchase quantity: "))

    if product_id not in inventory:
        raise ProductNotFoundError("Product does not exist.")

    if quantity <= 0:
        raise InvalidQuantityError("Quantity must be greater than zero.")

    if quantity > inventory[product_id]["stock"]:
        raise InsufficientStockError("Not enough stock available.")

except ProductNotFoundError as e:
    print("Product Error:", e)

except InvalidQuantityError as e:
    print("Quantity Error:", e)

except InsufficientStockError as e:
    print("Stock Error:", e)

except ValueError:
    print("Invalid input! Enter a whole number.")

else:
    inventory[product_id]["stock"] -= quantity

    print("\nPurchase successful!")
    print("Product:", inventory[product_id]["name"])
    print("Purchased Quantity:", quantity)
    print("Remaining Stock:", inventory[product_id]["stock"])

finally:
    print("\n----- INVENTORY AUDIT -----")
    print("Purchase attempt finished.")
    print("Current Inventory:")

    for product_id, details in inventory.items():
        print(product_id, details["name"], "Stock:", details["stock"])
```

### Q30. Mini Banking System with a Custom Exception Hierarchy

Design a menu-based mini banking program that supports deposit, withdrawal, and balance viewing. Deposits and withdrawals must be positive. A withdrawal cannot exceed the available balance. Invalid menu choices must also be handled.

Create a base custom exception named `BankingError`, then create `InvalidAmountError` and `InsufficientBalanceError` as child classes of `BankingError`. Raise the appropriate exception from the relevant operation and handle the base class or specific child classes safely. Keep the program running after an invalid transaction, and show the final account balance when the user chooses to exit.

```python
class BankingError(Exception):
    pass


class InvalidAmountError(BankingError):
    pass


class InsufficientBalanceError(BankingError):
    pass


balance = 10000

while True:

    print("\n===== MINI BANKING SYSTEM =====")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. View Balance")
    print("4. Exit")

    try:
        choice = int(input("Enter your choice: "))

        if choice == 1:

            amount = float(input("Enter deposit amount: "))

            if amount <= 0:
                raise InvalidAmountError("Deposit amount must be positive.")

            balance += amount

            print("Deposit successful!")

        elif choice == 2:

            amount = float(input("Enter withdrawal amount: "))

            if amount <= 0:
                raise InvalidAmountError("Withdrawal amount must be positive.")

            if amount > balance:
                raise InsufficientBalanceError("Insufficient account balance.")

            balance -= amount

            print("Withdrawal successful!")

        elif choice == 3:

            print("Current Balance:", balance)

        elif choice == 4:

            print("Thank you for using our banking system!")
            break

        else:
            print("Invalid choice! Please select between 1 and 4.")

    except InvalidAmountError as e:
        print("Amount Error:", e)

    except InsufficientBalanceError as e:
        print("Balance Error:", e)

    except ValueError:
        print("Invalid input! Please enter a valid number.")

    print("Current Account Balance:", balance)

print("\nFinal Account Balance:", balance)
```
