## Question 16 — Student Marks Processor

**Problem:**  
A teacher has a list of marks of all students in a class and wants key statistics.

**Requirements:**
- Create a function that accepts a **list of student marks**.
- Using a loop (or appropriate logic), calculate and **return**:
  - Highest mark
  - Lowest mark
  - Total marks
  - Average marks
- Display the results outside the function.

**Focus:** Function + list + loop + `return`

````python
def marks_processor(marks):
    highest = marks[0]
    lowest = marks[0]
    total = 0

    for mark in marks:
        if mark > highest:
            highest = mark

        if mark < lowest:
            lowest = mark

        total += mark

    average = total / len(marks)

    return highest, lowest, total, average

marks = [85, 90, 65, 78, 95, 72]

highest, lowest, total, average = marks_processor(marks)

print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
print("Total Marks:", total)
print("Average Marks:", average)
````

## Question 23 — Nested Salary Calculator

**Problem:**  
Salary calculation has several smaller steps. Use nested functions to organize the logic cleanly.

**Requirements:**
- Create an outer function called `calculate_salary()`.
- Inside it, create separate **nested functions** for:
  - Calculating allowance
  - Calculating deduction
  - Calculating final salary
- The outer function should combine the results and return the final salary.
- Call the outer function and display the result.

**Focus:** Nested functions

````python
def calculate_salary(basic_salary):

    def calculate_allowance():
        return basic_salary * 0.20

    def calculate_deduction():
        return basic_salary * 0.10

    def final_salary():
        allowance = calculate_allowance()
        deduction = calculate_deduction()

        return basic_salary + allowance - deduction

    return final_salary()

salary = calculate_salary(50000)

print("Final Salary: Rs.", salary)
````

## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
def process_order(price, quantity, discount_percent, tax_percent):

    def calculate_subtotal():
        return price * quantity

    def calculate_discount():
        return calculate_subtotal() * discount_percent / 100

    def calculate_tax():
        amount = calculate_subtotal() - calculate_discount()
        return amount * tax_percent / 100

    def calculate_final():
        subtotal = calculate_subtotal()
        discount = calculate_discount()
        tax = calculate_tax()

        return subtotal - discount + tax

    return calculate_final()

final_amount = process_order(1000, 3, 10, 5)

print("Final Order Amount: Rs.", final_amount)
````

````python
def customer_info(**kwargs):
    return kwargs


def product_details():
    products = {
        "Keyboard": 2500,
        "Mouse": 1500,
        "Headphones": 3000,
        "Monitor": 25000
    }

    return products

def calculate_product_total(price, quantity):
    return price * quantity

def calculate_subtotal(*args):
    return sum(args)

def calculate_discount(subtotal, discount_percent=10):
    if subtotal >= 5000:
        return subtotal * discount_percent / 100
    else:
        return 0

def calculate_tax(amount):
    return amount * 5 / 100

def delivery_charges(subtotal):
    if subtotal >= 10000:
        return 0
    else:
        return 200

def process_order(subtotal, discount, tax, delivery):

    def calculate_after_discount():
        return subtotal - discount

    def calculate_final_amount():
        amount = calculate_after_discount()
        return amount + tax + delivery

    return calculate_final_amount()

def order_summary(customer, cart, subtotal, discount, tax, delivery, total):

    print("\n========== ORDER SUMMARY ==========")

    print("Customer Name:", customer["name"])
    print("Phone:", customer["phone"])
    print("-----------------------------------")

    for item in cart:
        print(item["name"], "x", item["quantity"], "=", item["total"])

    print("-----------------------------------")
    print("Subtotal: Rs.", subtotal)
    print("Discount: Rs.", discount)
    print("Tax: Rs.", tax)
    print("Delivery Charges: Rs.", delivery)
    print("-----------------------------------")
    print("Final Amount: Rs.", total)
    print("===================================")


customer = customer_info(
    name="Ali",
    phone="03001234567"
)

products = product_details()

cart = []

print("========== AVAILABLE PRODUCTS ==========")

for name, price in products.items():
    print(name, ":", price)

print("----------------------------------------")


valid_quantity = lambda quantity: quantity > 0

while True:

    product_name = input("Enter product name (or done to finish): ").title()

    if product_name == "Done":
        break

    if product_name not in products:
        print("Product not available!")
        continue

    quantity = int(input("Enter quantity: "))

    if not valid_quantity(quantity):
        print("Quantity must be greater than zero!")
        continue

    price = products[product_name]

    total = calculate_product_total(price, quantity)

    cart.append({
        "name": product_name,
        "price": price,
        "quantity": quantity,
        "total": total
    })

    print("Product added successfully!")


if len(cart) == 0:
    print("Your shopping cart is empty.")

else:

    subtotal = calculate_subtotal(
        *[item["total"] for item in cart]
    )

    discount = calculate_discount(subtotal)

    after_discount = subtotal - discount
    tax = calculate_tax(after_discount)

    delivery = delivery_charges(subtotal)

    total = process_order(subtotal, discount, tax, delivery)

    order_summary(
        customer,
        cart,
        subtotal,
        discount,
        tax,
        delivery,
        total
    )
````
